```python
from typing import Callable

# Define a function that matches the Callable[[str, float, float], None] signature
def print_item_details(name: str, price: float, tax_rate: float) -> None:
    total = price + (price * tax_rate)
    print(f"{name}: £{total:.2f}")

# Define a function that accepts a Callable as an argument
def process_item(item_name: str, price: float, tax: float, callback: Callable[[str, float, float], None]) -> None:
    callback(item_name, price, tax)

# Usage
process_item("Laptop", 999.00, 0.20, print_item_details)
```
